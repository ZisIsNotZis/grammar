```python
from torch.nn import *
import torch
import numpy as np
from tqdm.notebook import tqdm
class F(Module):
    def __init__(s, a):
        super().__init__()
        s.a = a
    def forward(s, x):
        return s.a(x)
class EncDec(Module):
    def __init__(s, n, a, b, c=Identity, d=Identity):
        super().__init__()
        s.a = ModuleList(a()for n in range(n))
        s.b = ModuleList(b()for n in range(n))
        s.c = ModuleList(c()for n in range(n))
        s.d = ModuleList(d()for n in range(n))
    def forward(s, x):
        a = []
        for i in s.a:
            a.append(x)
            x = i(x)
        for i,j,k,a in zip(s.b, s.c, s.d, a[::-1]):
            x = k(i(x).add_(j(a)))
        return x
class Add(Sequential):
    def forward(s, x):
        t = s[0](x)
        for s in s[1:]:
            t += s(x)
        return t
class Mul(Sequential):
    def forward(s, x):
        return s[0](x).mul_(s[1](x))
torch.set_default_tensor_type(torch.cuda.FloatTensor)
a = np.memmap('../dat/py/all.90', mode='r')
a = torch.tensor(a).as_strided([len(a)-511,512], [1,1])
```

```python
m = 16
m = Sequential(
    Embedding(90, m),
    F(lambda x:x.transpose(1,2)),
    EncDec(3,
        lambda:Sequential(
            Add(
                Conv1d(m, m, 3, padding=1, bias=False),
                Sequential(
                    BatchNorm1d(m, track_running_stats=False, affine=False),
                    Mul(
                        Sequential(
                            Conv1d(m, m, 1, bias=False),
                            SiLU(True),
                        ),
                        Conv1d(m, m, 1, bias=False),
                    ),
                    Conv1d(m, m, 1, bias=False),
                ),
                Identity(),
            ),
            Add(
                Conv1d(m, m, 3, padding=1, bias=False),
                Sequential(
                    BatchNorm1d(m, track_running_stats=False, affine=False),
                    Mul(
                        Sequential(
                            Conv1d(m, m, 1, bias=False),
                            SiLU(True),
                        ),
                        Conv1d(m, m, 1, bias=False),
                    ),
                    Conv1d(m, m, 1, bias=False),
                ),
                Identity(),
            ),
            MaxPool1d(2),
        ),
        lambda:Sequential(
            F(lambda x:x.repeat_interleave(2,2)),
            Add(
                Conv1d(m, m, 3, padding=1, bias=False),
                Sequential(
                    BatchNorm1d(m, track_running_stats=False, affine=False),
                    Mul(
                        Sequential(
                            Conv1d(m, m, 1, bias=False),
                            SiLU(True),
                        ),
                        Conv1d(m, m, 1, bias=False),
                    ),
                    Conv1d(m, m, 1, bias=False),
                ),
                Identity(),
            ),
            Add(
                Conv1d(m, m, 3, padding=1, bias=False),
                Sequential(
                    BatchNorm1d(m, track_running_stats=False, affine=False),
                    Mul(
                        Sequential(
                            Conv1d(m, m, 1, bias=False),
                            SiLU(True),
                        ),
                        Conv1d(m, m, 1, bias=False),
                    ),
                    Conv1d(m, m, 1, bias=False),
                ),
                Identity(),
            ),
        ),
    ),
    F(lambda x:functional.conv1d(x,m[0].weight[:,:,None]))
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, .005, len(a)*5//(512*a.size(1)), pct_start=.01, div_factor=9, final_div_factor=99)
for x in tqdm(range(O.total_steps//200)):
    for x in range(200):
        y = a[torch.randint(len(a), [512])].long()
        x = m(y.put(torch.randint(y.numel(), [y.numel()*3//4]), torch.randint(90, [y.numel()*3//4])))
        functional.cross_entropy(x, y).backward()
        o.step()
        o.zero_grad()
        O.step()
    print((x.argmax(1)==y).sum()/y.numel(), end=' ')
```
