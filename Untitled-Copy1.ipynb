```python
import torch
from torch.nn import *
import numpy as np

class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a
    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)

class EncDec(Sequential):
    def forward(s, x):
        a = []
        while x.size(2) > 1:
            a.append(x)
            x = s[0](x)
        x = s[1](x)
        for a in a[::-1]:
            x = s[2](x) + s[1](a)
        return x

class Add(Sequential):
    def forward(s, x):
        return s[0](x) + s[1](x)

torch.set_default_tensor_type(torch.cuda.FloatTensor)
a = np.memmap('../dat/py/all.90', mode='r')
a = torch.tensor(a).as_strided([len(a)-7,8], [1,1])
```

```python
class Enc2(Module):
    def __init__(s, m):
        super().__init__()
        s.a = Sequential(
            BatchNorm1d(m, track_running_stats=0),
            ReLU(1),
        )
        s.b = Sequential(
            BatchNorm1d(m, track_running_stats=0),
            ReLU(1),
        )
        s._a = Conv1d(m, m, 3, padding=1, bias=0)
        s._b = Conv1d(m, m, 3, padding=1, bias=0)
        s.__ = Conv1d(m, m, 1, bias=0)
        s.ab = Conv1d(m, m, 3, padding=1, bias=0)
        s.a_ = Conv1d(m, m, 1, bias=0)
        s.b_ = Conv1d(m, m, 1, bias=0)
    def forward(s, x):
        t = s.a(s._a(x))
        return torch.max_pool1d(s.b_(s.b(s.ab(t)+s._b(x)))+s.a_(t)+s.__(x), 2)
class Dec2(Module):
    def __init__(s, m):
        super().__init__()
        s.a = Sequential(
            BatchNorm1d(m, track_running_stats=0),
            ReLU(1),
        )
        s.b = Sequential(
            BatchNorm1d(m, track_running_stats=0),
            ReLU(1),
        )
        s._a = ConvTranspose1d(m, m, 3, 2, 1, 1, bias=0)
        s._b = ConvTranspose1d(m, m, 3, 2, 1, 1, bias=0)
        s.__ = ConvTranspose1d(m, m, 3, 2, 1, 1, bias=0)
        s.ab = Conv1d(m, m, 3, padding=1, bias=0)
        s.a_ = Conv1d(m, m, 1, bias=0)
    def forward(s, x):
        t = s.a(s._a(x))
        return s.b(s.ab(t)+s._b(x))+s.a_(t)+s.__(x)
```

```python
from tqdm.notebook import tqdm
m = 32
m = Sequential(
    Embedding(90, m),
    F(lambda x: x.transpose(1,2)),
    EncDec(
        Enc2(m),
        Add(
            Sequential(
                Conv1d(m, m, 1, bias=0),
                BatchNorm1d(m, track_running_stats=0),
                ReLU(1),
                Conv1d(m, m, 1, bias=0),
            ),
            Conv1d(m, m, 1),
        ),
        Dec2(m),
    ),
    Add(
        Sequential(
            Conv1d(m, m, 1, bias=0),
            BatchNorm1d(m, track_running_stats=0),
            ReLU(1),
            Conv1d(m, 90, 1, bias=0),
        ),
        Conv1d(m, 90, 1),
    ),
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, .01, len(a)//81920, pct_start=.1, div_factor=9, final_div_factor=99)
I = torch.arange(8192)[:,None]
for x in tqdm(range(O.total_steps//200)):
    for x in range(200):
        y = a[torch.randint(len(a), [8192])].long()
        x = m(y.put(torch.randint(65536, [32768]), torch.randint(90, [32768])))
        functional.cross_entropy(x, y).backward()
        o.step()
        o.zero_grad()
        O.step()
    print((x.argmax(1)==y).sum()/y.numel())
```
