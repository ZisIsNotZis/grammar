```python
from torch.nn import *
import torch
import numpy as np
from tqdm.notebook import tqdm
class F(Module):
    def __init__(s, f, *a):
        super().__init__()
        s.f = f
        s.a = a
    def forward(s, x):
        return s.f(x, *s.a)
    def extra_repr(s):
        return ', '.join(map(str, (s.f,)+s.a))
class Add(Sequential):
    def forward(s, x):
        t = s[0](x)
        for i in s[1:]:
            t += i(x)
        return t
class Mul(Sequential):
    def forward(s, x):
        t = s[0](x)
        for i in s[1:]:
            t *= i(x)
        return t
class Res(Sequential):
    def forward(s, x):
        return super().forward(x).add_(x)
def EncDec(n, a, b):
    m = Res(*a(), *b())
    for n in range(n-1):
        m = Res(*a(), m, *b())
    return m
def Nn(i, o=0, b=False):
    return Sequential(
        Conv1d(i, i, 1, bias=False),
        BatchNorm1d(i, track_running_stats=False),
        SiLU(True),
        Conv1d(i, o or i, 1, bias=b),
    )
def SwiGLU(i, o=0, b=False):
    return Res(
        BatchNorm1d(i, affine=False, track_running_stats=False),
        Mul(
            Sequential(
                Conv1d(i, i, 1),
                SiLU(True)
            ),
            Conv1d(i, i, 1),
        ),
        Conv1d(i, o or i, 1, bias=b),
    )
```

```python
m = 16
m = Sequential(
    Embedding(90, m),
    F(lambda x:x.transpose(1,2)),
    EncDec(9,
        lambda:Sequential(
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            SwiGLU(m),
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            SwiGLU(m),
            MaxPool1d(2),
        ),
        lambda:Sequential(
            SwiGLU(m),
            ConvTranspose1d(m, m, 3, 2, 1, 1, m, False),
            SwiGLU(m),
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            Conv1d(m, m, 1, bias=False),
        ),
    ),
    SwiGLU(m, 90, True),
)
```
