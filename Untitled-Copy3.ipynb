```python
import torch
from torch.nn import *
import numpy as np

class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a
    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)

class EncDec(Module):
    def __init__(s, n, a, b, c):
        super().__init__()
        s.a = ModuleList(a()for n in range(n))
        s.b = ModuleList(b()for n in range(n))
        s.c = ModuleList(c()for n in range(n))
    def forward(s, x):
        a = []
        for i in s.a:
            a.append(x)
            x = i(x)
        for i, j, a in zip(s.b, s.c, a[::-1]):
            x = i(a) + j(x)
        return x

class Add(Sequential):
    def forward(s, x):
        return s[0](x) + s[1](x)

torch.set_default_tensor_type(torch.cuda.FloatTensor)
a = np.memmap('../dat/py/all.90', mode='r')
a = torch.tensor(a).as_strided([len(a)-255,256], [1,1])
```

```python
from tqdm.notebook import tqdm
m = 256
m = Sequential(
    Embedding(90, m),
    F(lambda x: x.transpose(1,2)),
    EncDec(3,
        lambda: Sequential(
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            Conv1d(m, m, 1),
            MaxPool1d(2),
            ReLU(1),
        ),
        Identity,
        lambda: Sequential(
            ConvTranspose1d(m, m, 3, 2, 1, 1, m, False),
            Conv1d(m, m, 1),
            ReLU(1),
        ),
    ),
    Conv1d(m, m, 1),
    ReLU(1),
    Conv1d(m, 90, 1),
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, .005, len(a)//(128*a.size(1)), pct_start=.1, div_factor=9, final_div_factor=99)
for x in tqdm(range(O.total_steps//100)):
    for x in range(100):
        y = a[torch.randint(len(a), [128])].long()
        x = m(y.put(torch.randint(y.numel(), [y.numel()*3//4]), torch.randint(90, [y.numel()*3//4])))
        functional.cross_entropy(x, y).backward()
        o.step()
        o.zero_grad()
        O.step()
    print((x.argmax(1)==y).sum()/y.numel())
```
