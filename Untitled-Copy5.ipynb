```python
import torch
from torch.nn import *
import numpy as np
from tqdm.notebook import tqdm

class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a
    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)
    
class EncDec(Module):
    def __init__(s, n, a, b, c=Identity):
        super().__init__()
        s.a = ModuleList(a()for n in range(n))
        s.b = ModuleList(b()for n in range(n))
        s.c = ModuleList(c()for n in range(n))
    def forward(s, x):
        a = []
        for i in s.a:
            a.append(x)
            x = i(x)
        for i,j,a in zip(s.b, s.c, a[::-1]):
            x = j(i(x) + a)
        return x

class Add(Sequential):
    def forward(s, x):
        return s[0](x) + s[1](x)

class Res(Sequential):
    def forward(s, x):
        return super().forward(x) + x

a = np.memmap('../dat/py/all.90', mode='r')
a = torch.from_numpy(a).as_strided([len(a)-31,32], [1,1])
```

```python
m = 13
m = Sequential(
    Embedding(90, m),
    F(lambda x:x.transpose(1,2)),
    EncDec(5,
        lambda:Sequential(
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            Res(
                Conv1d(m, m, 1, bias=False),
                BatchNorm1d(m, track_running_stats=False),
                SiLU(1),
                Conv1d(m, m, 1, bias=False),
            ),
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            Res(
                Conv1d(m, m, 1, bias=False),
                BatchNorm1d(m, track_running_stats=False),
                SiLU(1),
                Conv1d(m, m, 1, bias=False),
            ),
            MaxPool1d(2),
        ),
        lambda:Sequential(
            ConvTranspose1d(m, m, 3, 2, 1, 1, m, False),
            Res(
                Conv1d(m, m, 1, bias=False),
                BatchNorm1d(m, track_running_stats=False),
                SiLU(1),
                Conv1d(m, m, 1, bias=False),
            ),
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            Res(
                Conv1d(m, m, 1, bias=False),
                BatchNorm1d(m, track_running_stats=False),
                SiLU(1),
                Conv1d(m, m, 1, bias=False),
            ),
        ),
        lambda:Res(
            Conv1d(m, m, 1, bias=False),
            BatchNorm1d(m, track_running_stats=False),
            SiLU(1),
            Conv1d(m, m, 1, bias=False),
        ),
    ),
    Conv1d(m, m, 1, bias=False),
    BatchNorm1d(m, track_running_stats=False),
    SiLU(1),
    Conv1d(m, 90, 1),
).cuda()
print(sum(m.numel()for m in m.parameters()if m.ndim>1))
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, .01, len(a)*2//(4096*a.size(1)), pct_start=.1, div_factor=9, final_div_factor=99)
for x in tqdm(range(O.total_steps//100)):
    for x in range(100):
        y = a[torch.randint(len(a), [4096])].to('cuda', torch.long)
        x = m(y.put(torch.randint(y.numel(), [y.numel()//2], device='cuda'), torch.randint(90, [y.numel()//2], device='cuda')))
        functional.cross_entropy(x, y).backward()
        o.step()
        o.zero_grad()
        O.step()
    print(int((x.argmax(1)==y).sum()*1000/y.numel()), end=' ')
```
