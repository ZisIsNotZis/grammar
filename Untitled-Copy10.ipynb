```python
from torch.nn import *
import torch
import numpy as np
from functools import reduce, partial
from tqdm.notebook import tqdm
class F(Module):
    def __init__(s, f):
        super().__init__()
        s.f = f
    def forward(s, x):
        return s.f(x)
    def extra_repr(s):
        return str(s.f)
class Add(Sequential):
    f = torch.Tensor.add_
    def forward(s, x):
        return reduce(type(s).f, (s(x)for s in s))
class Mul(Add):
    f = torch.Tensor.mul_
class Res(Sequential):
    def forward(s, x):
        return super().forward(x).add_(x)
def Rec(n, f):
    m = None
    for n in range(n):
        m = f(m)
    return m
def EncDec(n, a, b):
    m = Res(*a(), *b())
    for n in range(n-1):
        m = Res(*a(), m, *b())
    return m
def Nn(i, o=0, b=False, h=1, a=ReLU):
    return Sequential(
        Conv1d(i, i*h, 1, bias=False),
        BatchNorm1d(i*h, track_running_stats=False),
        a(True),
        Conv1d(i*h, o or i, 1, bias=b),
    )
def SwiGLU(i, o=0, b=False, h=1, a=Hardswish):
    return Sequential(
        BatchNorm1d(i, affine=False, track_running_stats=False),
        Mul(
            Sequential(
                Conv1d(i, i*h, 1),
                a(True),
            ),
            Conv1d(i, i*h, 1),
        ),
        Conv1d(i*h, o or i, 1, bias=b),
    )
torch.set_default_tensor_type(torch.cuda.FloatTensor)
a = np.memmap('../dat/py/all.90', mode='r')
a = torch.tensor(a).as_strided((len(a)-511,512), (1,1))
```
