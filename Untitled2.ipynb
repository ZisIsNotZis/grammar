```python
import torch
from torch.nn import *
import numpy as np
from tqdm.notebook import tqdm

class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a
    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)
    
class EncDec(Module):
    def __init__(s, n, a, b, c):
        super().__init__()
        s.a = ModuleList(a()for n in range(n))
        s.b = ModuleList(b()for n in range(n))
        s.c = ModuleList(c()for n in range(n))
    def forward(s, x):
        a = []
        for i in s.a:
            a.append(x)
            x = i(x)
        for i,j,a in zip(s.b, s.c, a[::-1]):
            x = j(i(x) + a)
        return x

class Add(Sequential):
    def forward(s, x):
        return s[0](x) + s[1](x)

class Res(Sequential):
    def forward(s, x):
        return super().forward(x) + x
```

```python
m = 96
m = Sequential(
    Embedding(90, m),
    F(lambda x:x.transpose(1,2)),
    EncDec(5,
        lambda:Sequential(
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            Add(
                Sequential(
                    Conv1d(m, m, 1, bias=False),
                    BatchNorm1d(m, track_running_stats=False),
                    SiLU(1),
                    Conv1d(m, m, 1, bias=False),
                ),
                Conv1d(m, m, 1, bias=False),
            ),
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            Add(
                Sequential(
                    Conv1d(m, m, 1, bias=False),
                    BatchNorm1d(m, track_running_stats=False),
                    SiLU(1),
                    Conv1d(m, m, 1, bias=False),
                ),
                Conv1d(m, m, 1, bias=False),
            ),
            MaxPool1d(2),
        ),
        lambda:Sequential(
            Add(
                Sequential(
                    Conv1d(m, m, 1, bias=False),
                    BatchNorm1d(m, track_running_stats=False),
                    SiLU(1),
                    Conv1d(m, m, 1, bias=False),
                ),
                Conv1d(m, m, 1, bias=False),
            ),
            ConvTranspose1d(m, m, 3, 2, 1, 1, m, False),
            Add(
                Sequential(
                    Conv1d(m, m, 1, bias=False),
                    BatchNorm1d(m, track_running_stats=False),
                    SiLU(1),
                    Conv1d(m, m, 1, bias=False),
                ),
                Conv1d(m, m, 1, bias=False),
            ),
            Conv1d(m, m, 3, padding=1, groups=m, bias=False),
            Add(
                Sequential(
                    Conv1d(m, m, 1, bias=False),
                    BatchNorm1d(m, track_running_stats=False),
                    SiLU(1),
                    Conv1d(m, m, 1, bias=False),
                ),
                Conv1d(m, m, 1, bias=False),
            ),
        ),
        Identity,
    ),
    Add(
        Sequential(
            Conv1d(m, m, 1, bias=False),
            BatchNorm1d(m, track_running_stats=False),
            SiLU(1),
            Conv1d(m, 90, 1, bias=False),
        ),
        Conv1d(m, 90, 1),
    ),
)
print(sum(m.numel()for m in m.parameters()if m.ndim>1))
```
