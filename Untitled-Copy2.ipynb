```python
import torch
from torch.nn import *
import numpy as np
from tqdm.notebook import tqdm

class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a
    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)
    
class EncDec(Module):
    def __init__(s, n, a, b, c=Identity):
        super().__init__()
        s.a = ModuleList(a()for n in range(n))
        s.b = ModuleList(b()for n in range(n))
        s.c = ModuleList(c()for n in range(n))
    def forward(s, x):
        a = []
        for i in s.a:
            a.append(x)
            x = i(x)
        for i,j,a in zip(s.b, s.c, a[::-1]):
            x = j(i(x) + a)
        return x

class Add(Sequential):
    def forward(s, x):
        return s[0](x) + s[1](x)

class Res(Sequential):
    def forward(s, x):
        return super().forward(x) + x

torch.set_default_tensor_type(torch.cuda.FloatTensor)
a = np.memmap('../dat/py/all.90', mode='r')
a = torch.tensor(a).as_strided([len(a)-31,32], [1,1])
```
