```python
import torch
from torch.nn import *
import numpy as np

class F(Module):
    def __init__(s, f, a=None):
        super().__init__()
        s.f = f
        s.a = a
    def forward(s, x):
        return s.f(x) if s.a is None else s.f(x, s.a)

class EncDec(Sequential):
    def forward(s, x):
        a = []
        while x.size(2) > 1:
            a.append(x)
            x = s[0](x)
        x = s[1](x)
        for a in a[::-1]:
            x = s[2](x) + s[1](a)
        return x

class Add(Sequential):
    def forward(s, x):
        return s[0](x) + s[1](x)

torch.set_default_tensor_type(torch.cuda.FloatTensor)
a = np.memmap('../dat/py/all.90', mode='r')
a = torch.tensor(a).as_strided([len(a)-63,64], [1,1])
```

```python
from tqdm.notebook import tqdm
m = 128
m = Sequential(
    Embedding(90, m),
    F(lambda x: x.transpose(1,2)),
    EncDec(
        Sequential(
            Conv1d(m, m, 3, padding=1),
            MaxPool1d(2),
            ReLU(1),
        ),
        Sequential(
            Conv1d(m, m, 1),
            ReLU(1),
            Conv1d(m, m, 1),
        ),
        Sequential(
            ConvTranspose1d(m, m, 3, 2, 1, 1),
            ReLU(1),
        ),
    ),
    Conv1d(m, m, 1),
    ReLU(1),
    Conv1d(m, 90, 1),
)
o = torch.optim.Adam(m.parameters())
O = torch.optim.lr_scheduler.OneCycleLR(o, .01, len(a)//20480, pct_start=.1, div_factor=9, final_div_factor=99)
I = torch.arange(8192)[:,None]
for x in tqdm(range(O.total_steps//200)):
    for x in range(200):
        y = a[torch.randint(len(a), [2048])].long()
        x = m(y.put(torch.randint(y.numel(), [y.numel()//2]), torch.randint(90, [y.numel()//2])))
        functional.cross_entropy(x, y).backward()
        o.step()
        o.zero_grad()
        O.step()
    print((x.argmax(1)==y).sum()/y.numel())
```
