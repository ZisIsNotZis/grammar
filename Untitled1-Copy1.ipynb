```python
import torch
from torch.nn import *
A = Conv1d(16, 16, 3, bias=False).eval()
B = Sequential(
    Conv1d(16, 16, 3, groups=16, bias=False),
    Conv1d(16, 16, 1, bias=False),
)
o = torch.optim.Adam(B.parameters(), .02)
for x in range(10):
    for x in range(1000):
        x = torch.rand(64, 16, 16)
        x = functional.mse_loss(A(x), B(x))
        x.backward()
        o.step()
        o.zero_grad()
    print(x)
```
